importando pandas

```python
import pandas as pd
```

importando scikit-learn,treino e teste,metricas

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn import metrics
```

selecionar as features do csv original do irirs e carregar o dataset

```python
iris = pd.read_csv('iris.csv.xls')
nome_coluna = ['SepalLength','SepalWidth','PetalLength','PetalWidth','Name']
```

Testando se eu carreguei corretamente

```python
iris.head()
```

Dividir as features/colunas para ele usar

```python
features_coluna = ['SepalLength','SepalWidth','PetalLength','PetalWidth']
```

separar em X(vou usar para classificar) e Y(classe que quero descobrir)

```python
X = iris[features_coluna]
Y = iris['Name']
```

agora separar de acordo com a proporção de treino e teste (em 70% treino,30% teste)

```python
X_treino, X_teste, Y_treino, Y_teste = train_test_split(X,Y, test_size=0.30, random_state=1)
```

chamar o classificador de árvores de decisão

```python
arvore = DecisionTreeClassifier()
```

encontro de fit arvore

```python
arvore = arvore.fit(X_treino, Y_treino)
```

Partes finais, tentando prever o resultado baseado na separação 70/30 e estudo dos 70%

```python
Y_pred = arvore.predict(X_teste)
```

Comparar o valor real (Y_teste) com a minha predição (valor predito, Y_pred)
Acurácia = Previsões Corretas/Total de Previsões

```python
print('Acuracia: ', metrics.accuracy_score(Y_teste, Y_pred))
```
