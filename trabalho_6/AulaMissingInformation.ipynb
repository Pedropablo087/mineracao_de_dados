Importando Pandas e printando a soma de nulls

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.impute import KNNImputer

df = pd.read_csv('winedataset_modified.csv')
print(df.isnull().sum())
```

Selecionando linhas com null

```python
df_filtra_null = df[df.isnull().any(axis=1)]
print(df_filtra_null)
```

Esquecendo linhas com valores null

```python
df_drop_linhas = df.dropna()
#print(df_drop_linhas.isnull().sum())
#print(df_drop_linhas)
```

Esquecendo colunas com valores null

```python
df_drop_colunas = df.drop(columns=['type','fixed acidity','volatile acidity','citric acid','residual sugar','chlorides','pH','sulphates'])
print(df_drop_colunas.isnull().sum())
print(df_drop_colunas)
```

Exibindo os detalhes do dataframe original, dataframe sem linhas com null, sem colunas com null

```python
print("Dataframe original", df.shape)
print("Dataframe original, sem linha com null", df_drop_linhas.shape)
print("Dataframe original, sem colunas com null", df_drop_colunas.shape)
```

```python
df.to_csv('original.csv', index=False)
df_drop_linhas.to_csv('semLinha.csv', index=False)
df_drop_colunas.to_csv('semColuna.csv', index=False)
```

Trocando null por moda

```python
df[df.index==124]
```

```python
dfNovo = df
dfNovo['type'] = dfNovo['type'].fillna(dfNovo['type'].mode()[0])
dfNovo[dfNovo.index==124]
```

```python
df2 = df.drop(columns='type')
colunas_faltando = df2.columns[df2.isna().any()].tolist()
sem_colunas_faltando = list(set(df2.columns) - set(colunas_faltando))
sem_colunas_faltando
```

```python
for coluna in colunas_faltando:
    df_temp = df2.dropna()
    X = df_temp[sem_colunas_faltando]
    y = df_temp[coluna]

    regressaoLinear = LinearRegression()
    regressaoLinear.fit(X,y)

    df2.loc[df2[coluna].isna(), coluna] = regressaoLinear.predict(df2.loc[df[coluna].isna(), sem_colunas_faltando])
```
