```python
import pandas as pd
import numpy as np
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.ensemble import HistGradientBoostingRegressor

# 1. Carrega o dataset (certifique-se que o arquivo está no mesmo diretório)
df = pd.read_csv('employees_dataset_with_missing.csv.xls')

print("Valores nulos ANTES da imputação:")
print(df.isnull().sum())

# 2. Seleciona apenas colunas numéricas (todas são numéricas)
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
df_numeric = df[numeric_cols].copy()

# 3. Cria o imputador iterativo com HistGradientBoostingRegressor
#    Esse regressor aceita NaN nativamente (trata na divisão das árvores)
imputer = IterativeImputer(
    estimator=HistGradientBoostingRegressor(
        random_state=42,
        max_iter=100,
        early_stopping=False
    ),
    max_iter=10,
    random_state=42
)

# 4. Aplica a imputação (os NaN são substituídos pelos valores estimados)
df_imputed_array = imputer.fit_transform(df_numeric)

# 5. Recria o DataFrame com os valores imputados
df_imputed = pd.DataFrame(df_imputed_array, columns=numeric_cols, index=df.index)

# 6. Substitui as colunas originais pelos valores imputados
for col in numeric_cols:
    df[col] = df_imputed[col]

print("\nValores nulos DEPOIS da imputação:")
print(df.isnull().sum())

# 7. Salva o dataset imputado (opcional)
df.to_excel('employees_dataset_imputed.xlsx', index=False)

print("\nArquivo 'employees_dataset_imputed.xlsx' salvo com sucesso!")
```
